```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
from langgraph.checkpoint.memory import InMemorySaver
import time
```

```python
# define state
class CrashState(TypedDict):
    input: str
    step1: str
    step2: str
    step3: str
    
```

```python
def step1(state: CrashState):
    print("step1 Executed")
    return {"step1": "done" , "input": state["input"]}

def step2(state: CrashState):
    print("step2 hanging... Now manually interupt from the notebook tool bar")
    time.sleep(30) #simulate long running hang
    return {"step2": "done"}
def step3(state: CrashState):
    print("step3 Executed")
    return {"step3": "done"}
```

```python
#Build the graph
builder = StateGraph(CrashState)
builder.add_node("step1", step1)
builder.add_node("step2", step2)
builder.add_node("step3", step3)

#add edges
builder.add_edge(START, "step1")
builder.add_edge("step1", "step2")
builder.add_edge("step2", "step3")
builder.add_edge("step3", END)

checkponter = InMemorySaver()
graph = builder.compile(checkpointer=checkponter)
```

```python
graph
```

```python
config = {"configurable": {"thread_id": "1"}}
try:
    print("Running graph. Please manually interrupt the kernel to simulate a crash in step 2")
    graph.invoke({"input": "start"} , config=config)

except KeyboardInterrupt:
    print("kernel manually interrupted. Crash Simulated")
    
```

```python
graph.get_state(config)
```

```python
list(graph.get_state_history(config))
```

```python
#Re run to show fault tolerance resume
final_state = graph.invoke(None, config=config)
print("Final State: ", final_state)
```
